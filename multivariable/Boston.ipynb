# Viviendas en Boston

El Boston Housing Dataset es un conjunto de datos derivado de información recopilada por el Servicio de Censo de Estados Unidos sobre viviendas en el área de Boston, Massachusetts.

Variables principales

- CRIM: tasa de criminalidad per cápita por ciudad
- ZN: proporción de suelo residencial para lotes grandes
- INDUS: proporción de áreas industriales
- CHAS: variable dummy (1 si está cerca del río Charles, 0 si no)
- NOX: concentración de óxidos de nitrógeno
- RM: número promedio de habitaciones por vivienda
- AGE: proporción de casas construidas antes de 1940
- DIS: distancia a centros de empleo
- RAD: accesibilidad a carreteras principales
- TAX: tasa de impuestos a la propiedad
- PTRATIO: ratio alumno-profesor
- B: indicador relacionado con composición poblacional
- LSTAT: porcentaje de población de bajos ingresos
- MEDV: valor medio de la vivienda (variable objetivo)

```python
import pandas as pd

df = pd.read_csv("BostonHousing.csv")
df
```

```python
print("Dimensión de la tabla: %s renglones( datos), %s columnas (variables)" % df.shape) 

# Estadísticas descriptivas
estadisticas = df.describe()
print("Estadísticas descriptivas\n", estadisticas)
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt

fig, axs = plt.subplots( ncols = 7, nrows=2, figsize=(20,10))

axs = axs.flatten()

for i, col in enumerate(df.columns):
    sns.boxplot(y=df[col], ax=axs[i])

plt.show()
```

```python
df.corr().abs()
```

```python
import mdr
df = df[df["medv"] < 50]
X = df["dis"].values
Y = df["medv"].values

modelo = mdr.RegresionLinealSimple(X,Y)
modelo.plot_with_intervals()
```

```python
modelo.summary()
```
